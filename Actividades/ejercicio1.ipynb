## Problema: División de cuenta con propina
Crea un programa que pida el total de la cuenta de un restaurante,
el porcentaje de propina a dejar y el número de personas que pagarán. El programa
debe calcular el monto de la propina, el total a pagar con propina y cuánto le toca
pagar a cada persona (con dos decimales).


## EPS aplicado:
1. Solicitar el total de la cuenta.

2. Solicitar el porcentaje de propina que se desea agregar.

3. Solicitar el número de comensales (personas) que pagarán la cuenta.

4. Mostrar el resumen inicial con los datos capturados (total de la cuenta, porcentaje de propina y número de comensales).

5. Calcular el monto de la propina multiplicando el total de la cuenta por el porcentaje de propina y dividiéndolo entre 100.

6. Calcular el total con propina sumando el total original de la cuenta más el monto de la propina calculada.

7. Calcular la cuota individual dividiendo el total con propina entre el número de comensales.

8. Mostrar los resultados finales: el monto de la propina, el total acumulado con propina y la cantidad a pagar por persona.


```python
total_de_cuenta=float(input("Ingresa el total de la cuenta: "))

porcentaje_propina = int(input("Ahora indica qué porcentaje te gustaría agregar de propina: "))

comensales = int(input("Ingresa el número de personas que van a pagar la cuenta: "))

print("Total de la cuenta: $", total_de_cuenta)
print("Porcentaje de propina: %", porcentaje_propina)
print("Número de personas: ", comensales)
print("\n")

print("Propina: $", total_de_cuenta * porcentaje_propina / 100)
print("Total con propina: $", total_de_cuenta + (total_de_cuenta * porcentaje_propina / 100))
print("Pago por persona: $", (total_de_cuenta + (total_de_cuenta * porcentaje_propina / 100)) / comensales)



```

## Problema: Conversor de minutos a días, horas y minutos.
Crea un programa que pida una cantidad total de minutos (entero) y
la convierta a días, horas y minutos restantes. Utiliza los operadores de división
entera.


## EPS aplicado:
1. Solicitar al usuario la cantidad total de minutos (como número entero).

2. Mostrar la cantidad total de minutos ingresada por el usuario.

3. Calcular la cantidad de días dividiendo el total de minutos entre 1440 (usando división entera).

4. Calcular las horas restantes tomando el residuo de los minutos sobrantes del día (% 1440) y dividiéndolo entre 60 (usando división entera).

5. Calcular los minutos restantes obteniendo el residuo del total de minutos entre 60 (% 60).

6. Mostrar el resultado desglosado en días, horas y minutos.

```python
total_de_minutos=int(input("Ingresa el total de minutos: "))

print("Total de minutos: ", total_de_minutos)
print("\n")
print(total_de_minutos, " minutos = ", total_de_minutos // 1440, " dia/s, ", (total_de_minutos % 1440) // 60, " hora/s, ", total_de_minutos % 60, " minuto/s")


```

## Problema: Calificación final ponderada
Crea un programa que pida las calificaciones de tres parciales
(valores de 0 a 10) y calcule la calificación final considerando una ponderación de
**30%, 30% y 40%** respectivamente. Muestra el resultado con dos decimales.


## EPS aplicado:
1. Mostrar un mensaje de instrucción indicando el rango válido para las calificaciones (valores del 1 al 10).

2. Solicitar la primera calificación (como número decimal).

3. Solicitar la segunda calificación (como número decimal).

4. Solicitar la tercera calificación (como número decimal).

5. Calcular el promedio final multiplicando la primera calificación por 0.30, la segunda por 0.30, la tercera por 0.40 y sumando los tres resultados.

6. Mostrar el resultado final del promedio ponderado.


```python
print("Ingresa tus calificaciones con valores del 1 al 10")
calificacion1 = float(input("Ingrese la primera calificación: "))
calificacion2 = float(input("Ingrese la segunda calificación: "))
calificacion3 = float(input("Ingrese la tercera calificación: "))

promedio= (calificacion1 *  0.30) + (calificacion2 * 0.30) + (calificacion3 * 0.40)
print("El promedio final de tu trimestre es:", promedio)
```

## Problema: Conversor de moneda (MXN a USD y EUR)
Crea un programa que pida una cantidad en pesos mexicanos y los
tipos de cambio del dólar (USD) y del euro (EUR). Debe calcular y mostrar las
equivalencias redondeadas a dos decimales.


## EPS aplicado:
1. Solicitar la cantidad en pesos mexicanos (MXN) a convertir (como número entero).

2. Definir los tipos de cambio fijos para el dólar (USD a 17.0) y el euro (EUR a 19.6).

3. Mostrar la cantidad ingresada en MXN y las tasas de cambio de referencia.

4. Calcular la conversión a dólares dividiendo la cantidad en pesos entre el tipo de cambio del dólar.

5. Calcular la conversión a euros dividiendo la cantidad en pesos entre el tipo de cambio del euro.

6. Mostrar los resultados finales de la conversión en USD y EUR redondeados a dos decimales.

```python
cantidad = int(input("Ingrese una cantidad en pesos mexicanos para convertir a dólares/euros: "))

dolares = 17.0
euros = 19.6
print("Cantidad en MXN: ", cantidad)
print("Tipo de cambio USD: ", dolares)
print("Tipo de cambio EUR: ", euros)

dolares = cantidad / dolares
euros = cantidad / euros

print(" MXN equivalen a: ")
print(f"USD: {dolares:.2f}")
print(f"EUR: {euros:.2f}")
```
