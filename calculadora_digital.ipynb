## Problema: Calculadora de tiempo invertido.
Desarrollar una calculadora interactiva en Python que permita registrar el
tiempo diario (en horas o fracciones de hora) que una persona dedica a distintas
plataformas digitales: redes sociales, mensajería, servicios de streaming,
videojuegos, entre otras. El programa debe capturar los datos, procesarlos y
mostrar un resumen claro y ordenado de los resultados.

## Proceso y modelo EPS aplicado:

- 

```python
Nombre_usuario = input(str("Ingresa tu nombre de usuario: "))
print("Hola", Nombre_usuario, "bienvenido a la calculadora de tiempo invertido en tus actividades digitales.", "\n")

redes_sociales = float(input("Ingresa el tiempo que pasas en redes sociales (en horas): "))
mensajeria = float(input("Ingresa el tiempo que pasas en mensajería (en horas): "))
servicios_streaming = float(input("Ingresa el tiempo que pasas en servicios de streaming (en horas): "))
fotografia = float(input("Ingresa el tiempo que pasas en fotografía (en horas): "))
videojuegos = float(input("Ingresa el tiempo que pasas en videojuegos (en horas): "))

tiempo_total = (redes_sociales + mensajeria + servicios_streaming + fotografia + videojuegos)
print("El tiempo total invertido en tus actividades digitales es: ", tiempo_total, "\n")

porcentaje = (tiempo_total / 24) * 100
print("El porcentaje del día dedicado a estas actividades digitales es: ", porcentaje, "%")
```
